```python
import pandas as pd

train_df = pd.read_csv("/content/train.csv")
test_df  = pd.read_csv("/content/test.csv")

print(train_df.shape, test_df.shape)
print(train_df.columns)
```

```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score

from perforatedai import globals_perforatedai as GPA
from perforatedai import utils_perforatedai as UPA

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

ID_COL = "id"
TARGET_COL = "Heart Disease"

# ====== preprocessing (tabular) ======
y = train_df[TARGET_COL].map({"Presence": 1, "Absence": 0}).astype(np.int64).values
X_df = train_df.drop(columns=[ID_COL, TARGET_COL]).copy()
T_df = test_df.drop(columns=[ID_COL]).copy()

cat_cols = ["Sex","Chest pain type","FBS over 120","EKG results","Exercise angina","Slope of ST","Number of vessels fluro","Thallium"]
cat_cols = [c for c in cat_cols if c in X_df.columns]
num_cols = [c for c in X_df.columns if c not in cat_cols]

# categorical -> codes (shared mapping across train+test)
all_cat = pd.concat([X_df[cat_cols], T_df[cat_cols]], axis=0, ignore_index=True)
cat_maps = {}
for c in cat_cols:
    all_cat[c] = all_cat[c].astype("int64")
    # make contiguous
    uniq = np.sort(all_cat[c].unique())
    cat_maps[c] = {v:i for i,v in enumerate(uniq)}
    all_cat[c] = all_cat[c].map(cat_maps[c]).astype("int64")

X_cat = all_cat.iloc[:len(X_df)][cat_cols].values.astype(np.int64)
T_cat = all_cat.iloc[len(X_df):][cat_cols].values.astype(np.int64)

# numeric -> scaled
scaler = StandardScaler()
X_num = scaler.fit_transform(X_df[num_cols].values.astype(np.float32)).astype(np.float32)
T_num = scaler.transform(T_df[num_cols].values.astype(np.float32)).astype(np.float32)

cat_sizes = [int(all_cat[c].nunique()) for c in cat_cols]

# ====== dataset ======
class TabDS(Dataset):
    def __init__(self, X_num, X_cat, y=None):
        self.Xn = torch.from_numpy(X_num)
        self.Xc = torch.from_numpy(X_cat)
        self.y = None if y is None else torch.from_numpy(y.astype(np.float32))
    def __len__(self): return self.Xn.shape[0]
    def __getitem__(self, i):
        if self.y is None:
            return self.Xn[i], self.Xc[i]
        return self.Xn[i], self.Xc[i], self.y[i]

# ====== model (strong NN) ======
class TabNet(nn.Module):
    def __init__(self, n_num, cat_sizes, emb_dim=8):
        super().__init__()
        self.embs = nn.ModuleList([nn.Embedding(s, min(emb_dim, max(2, s//2))) for s in cat_sizes])
        cat_out = sum(e.embedding_dim for e in self.embs)

        self.bn0 = nn.BatchNorm1d(n_num + cat_out)
        self.fc1 = nn.Linear(n_num + cat_out, 256)
        self.bn1 = nn.BatchNorm1d(256)
        self.fc2 = nn.Linear(256, 128)
        self.bn2 = nn.BatchNorm1d(128)
        self.fc3 = nn.Linear(128, 64)
        self.out = nn.Linear(64, 1)

        self.drop = nn.Dropout(0.25)

    def forward(self, x_num, x_cat):
        embs = [emb(x_cat[:,i]) for i,emb in enumerate(self.embs)]
        x = torch.cat([x_num] + embs, dim=1)
        x = self.bn0(x)
        x = self.drop(F.relu(self.bn1(self.fc1(x))))
        x = self.drop(F.relu(self.bn2(self.fc2(x))))
        x = self.drop(F.relu(self.fc3(x)))
        return self.out(x).squeeze(-1)

def run_fold(train_idx, val_idx, max_outer=80):
    ds_tr = TabDS(X_num[train_idx], X_cat[train_idx], y[train_idx])
    ds_va = TabDS(X_num[val_idx], X_cat[val_idx], y[val_idx])
    dl_tr = DataLoader(ds_tr, batch_size=8192, shuffle=True, num_workers=0)
    dl_va = DataLoader(ds_va, batch_size=16384, shuffle=False, num_workers=0)

    base = TabNet(n_num=X_num.shape[1], cat_sizes=cat_sizes).to(device)

    # IMPORTANT: avoid debug/test behavior
    if hasattr(GPA, "pc") and hasattr(GPA.pc, "set_testing_dendrite_capacity"):
        GPA.pc.set_testing_dendrite_capacity(False)

    # add dendrites
    model = UPA.initialize_pai(base).to(device)

    opt = torch.optim.Adam(model.parameters(), lr=1e-3)
    if hasattr(GPA.pai_tracker, "set_optimizer_instance"):
        GPA.pai_tracker.set_optimizer_instance(opt)
    else:
        GPA.pai_tracker.setOptimizerInstance(opt)

    scaler_amp = torch.cuda.amp.GradScaler(enabled=torch.cuda.is_available())

    def train_steps(steps=400):
        model.train()
        it = iter(dl_tr)
        for _ in range(steps):
            try:
                xn, xc, yy = next(it)
            except StopIteration:
                it = iter(dl_tr)
                xn, xc, yy = next(it)
            xn, xc, yy = xn.to(device), xc.to(device), yy.to(device)
            opt.zero_grad(set_to_none=True)
            with torch.cuda.amp.autocast(enabled=torch.cuda.is_available()):
                logits = model(xn, xc)
                loss = F.binary_cross_entropy_with_logits(logits, yy)
            scaler_amp.scale(loss).backward()
            scaler_amp.step(opt)
            scaler_amp.update()

    @torch.no_grad()
    def val_auc():
        model.eval()
        ps=[]; ys=[]
        for xn, xc, yy in dl_va:
            xn, xc = xn.to(device), xc.to(device)
            logits = model(xn, xc)
            ps.append(torch.sigmoid(logits).cpu().numpy())
            ys.append(yy.numpy())
        p = np.concatenate(ps); t = np.concatenate(ys)
        return float(roc_auc_score(t, p)), p

    best_auc = 0.0
    best_val_pred = None

    # PAI outer loop
    for outer in range(max_outer):
        train_steps(steps=600)  # strong training per validation
        auc, vp = val_auc()
        if auc > best_auc:
            best_auc = auc
            best_val_pred = vp.copy()

        model, restructured, complete = GPA.pai_tracker.add_validation_score(auc, model)
        model = model.to(device)

        print(f"[PAI] outer={outer} val_auc={auc:.6f} best={best_auc:.6f} restr={restructured} complete={complete}")

        if complete:
            break
        if restructured:
            opt = torch.optim.Adam(model.parameters(), lr=1e-3)
            if hasattr(GPA.pai_tracker, "set_optimizer_instance"):
                GPA.pai_tracker.set_optimizer_instance(opt)
            else:
                GPA.pai_tracker.setOptimizerInstance(opt)

    return model, best_auc

# ====== Train 3-fold and predict ======
skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
test_pred = np.zeros(len(T_num), dtype=np.float32)
fold_aucs = []

for fold, (tr, va) in enumerate(skf.split(X_num, y), 1):
    print("\n=== Fold", fold, "===")
    model, best_auc = run_fold(tr, va, max_outer=60)
    fold_aucs.append(best_auc)

    # predict test
    ds_te = TabDS(T_num, T_cat, y=None)
    dl_te = DataLoader(ds_te, batch_size=16384, shuffle=False, num_workers=0)
    preds=[]
    model.eval()
    with torch.no_grad():
        for xn, xc in dl_te:
            xn, xc = xn.to(device), xc.to(device)
            preds.append(torch.sigmoid(model(xn, xc)).cpu().numpy())
    test_pred += (np.concatenate(preds).astype(np.float32) / 3.0)

print("Fold AUCs:", fold_aucs, "mean:", float(np.mean(fold_aucs)))

# safety: if model is inverted, flip once (AUC would be < 0.5 typically)
if np.mean(fold_aucs) < 0.5:
    print("AUC < 0.5 detected, flipping predictions")
    test_pred = 1.0 - test_pred

submission = pd.DataFrame({"id": test_df["id"].values, "Heart Disease": test_pred})
submission.to_csv("/content/submission.csv", index=False)
print("Saved /content/submission.csv")
print(submission.head())
```

```python
# =========================================================
# ~6 MINUTES Colab Run (T4): PerforatedAI Dendrites Tabular NN
# - 1-fold (StratifiedShuffleSplit)
# - hard stop: MAX_OUTER=10, TRAIN_STEPS=140
# - skips PAI interactive ipdb prompts
# - GPU + AMP
# - writes /content/submission.csv
# =========================================================

!pip -q install git+https://github.com/PerforatedAI/PerforatedAI.git
!pip -q install scikit-learn pandas numpy

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score

from perforatedai import globals_perforatedai as GPA
from perforatedai import utils_perforatedai as UPA

# ----------------------------
# Config (6-min budget)
# ----------------------------
TRAIN_PATH = "/content/train.csv"
TEST_PATH  = "/content/test.csv"
#OUT_PATH   = "/content/submission.csv"
OUT_PATH = "/content/submission_2024.csv"

ID_COL = "id"
TARGET_COL = "Heart Disease"

#SEED = 42
SEED = 2024

VAL_FRAC = 0.12
MAX_OUTER = 10              # hard cap for time
TRAIN_STEPS_PER_OUTER = 140 # hard cap for time
BATCH_TRAIN = 16384
BATCH_EVAL  = 32768

LR = 1e-3
# PerforatedAI recommends no weight decay; keep it 0 for speed + fewer prompts
WEIGHT_DECAY = 0.0

# Plateau stop: if best doesn't improve for N outers, exit
NO_IMPROVE_LIMIT = 4
IMPROVE_EPS = 3e-5

torch.manual_seed(SEED)
np.random.seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device, torch.cuda.get_device_name(0) if torch.cuda.is_available() else "")

# ----------------------------
# Load data
# ----------------------------
train_df = pd.read_csv(TRAIN_PATH)
test_df  = pd.read_csv(TEST_PATH)

y = train_df[TARGET_COL].map({"Presence": 1, "Absence": 0}).astype(np.int64).values
X_df = train_df.drop(columns=[ID_COL, TARGET_COL]).copy()
T_df = test_df.drop(columns=[ID_COL]).copy()

# Categorical columns (heart dataset standard)
cat_cols = ["Sex","Chest pain type","FBS over 120","EKG results","Exercise angina","Slope of ST","Number of vessels fluro","Thallium"]
cat_cols = [c for c in cat_cols if c in X_df.columns]
num_cols = [c for c in X_df.columns if c not in cat_cols]

# Encode categoricals jointly to keep consistent codes
if len(cat_cols) > 0:
    all_cat = pd.concat([X_df[cat_cols], T_df[cat_cols]], axis=0, ignore_index=True)
    for c in cat_cols:
        all_cat[c] = all_cat[c].astype("int64")
        uniq = np.sort(all_cat[c].unique())
        mp = {v:i for i,v in enumerate(uniq)}
        all_cat[c] = all_cat[c].map(mp).astype("int64")

    X_cat = all_cat.iloc[:len(X_df)][cat_cols].values.astype(np.int64)
    T_cat = all_cat.iloc[len(X_df):][cat_cols].values.astype(np.int64)
    cat_sizes = [int(all_cat[c].nunique()) for c in cat_cols]
else:
    X_cat = np.zeros((len(X_df), 0), dtype=np.int64)
    T_cat = np.zeros((len(T_df), 0), dtype=np.int64)
    cat_sizes = []

# Scale numerics
scaler = StandardScaler()
X_num = scaler.fit_transform(X_df[num_cols].values.astype(np.float32)).astype(np.float32)
T_num = scaler.transform(T_df[num_cols].values.astype(np.float32)).astype(np.float32)

# ----------------------------
# Dataset
# ----------------------------
class TabDS(Dataset):
    def __init__(self, Xn, Xc, y=None):
        self.Xn = torch.from_numpy(Xn)
        self.Xc = torch.from_numpy(Xc)
        self.y  = None if y is None else torch.from_numpy(y.astype(np.float32))
    def __len__(self): return self.Xn.shape[0]
    def __getitem__(self, i):
        if self.y is None:
            return self.Xn[i], self.Xc[i]
        return self.Xn[i], self.Xc[i], self.y[i]

# ----------------------------
# Strong but fast model
# ----------------------------
class TabModel(nn.Module):
    def __init__(self, n_num, cat_sizes, emb_dim=8, dropout=0.15):
        super().__init__()
        self.embs = nn.ModuleList([nn.Embedding(s, min(emb_dim, max(2, s//2))) for s in cat_sizes])
        cat_out = sum(e.embedding_dim for e in self.embs)

        self.bn0 = nn.BatchNorm1d(n_num + cat_out)
        self.fc1 = nn.Linear(n_num + cat_out, 256)
        self.bn1 = nn.BatchNorm1d(256)
        self.fc2 = nn.Linear(256, 128)
        self.bn2 = nn.BatchNorm1d(128)
        self.fc3 = nn.Linear(128, 64)
        self.out = nn.Linear(64, 1)
        self.drop = nn.Dropout(dropout)

    def forward(self, x_num, x_cat):
        if len(self.embs) > 0:
            embs = [emb(x_cat[:, i]) for i, emb in enumerate(self.embs)]
            x = torch.cat([x_num] + embs, dim=1)
        else:
            x = x_num
        x = self.bn0(x)
        x = self.drop(F.relu(self.bn1(self.fc1(x))))
        x = self.drop(F.relu(self.bn2(self.fc2(x))))
        x = self.drop(F.relu(self.fc3(x)))
        return self.out(x).squeeze(-1)

# ----------------------------
# One fast stratified split
# ----------------------------
sss = StratifiedShuffleSplit(n_splits=1, test_size=VAL_FRAC, random_state=SEED)
tr_idx, va_idx = next(sss.split(X_num, y))

ds_tr = TabDS(X_num[tr_idx], X_cat[tr_idx], y[tr_idx])
ds_va = TabDS(X_num[va_idx], X_cat[va_idx], y[va_idx])

dl_tr = DataLoader(ds_tr, batch_size=BATCH_TRAIN, shuffle=True, num_workers=0)
dl_va = DataLoader(ds_va, batch_size=BATCH_EVAL, shuffle=False, num_workers=0)

# ----------------------------
# PerforatedAI config: avoid prompts
# ----------------------------
# Skip "unwrapped modules" confirmations (BN + Embeddings)
if hasattr(GPA, "pc"):
    if hasattr(GPA.pc, "set_unwrapped_modules_confirmed"):
        GPA.pc.set_unwrapped_modules_confirmed(True)
    # Track unwrapped modules so PAI won't complain
    try:
        GPA.pc.get_module_names_to_track().add("BatchNorm1d")
        GPA.pc.get_module_names_to_track().add("Embedding")
    except Exception:
        pass

    # If you keep weight_decay=0, no warning; but accept anyway to avoid breakpoints
    if hasattr(GPA.pc, "set_weight_decay_accepted"):
        GPA.pc.set_weight_decay_accepted(True)

    # Disable test/debug capacity behavior
    if hasattr(GPA.pc, "set_testing_dendrite_capacity"):
        GPA.pc.set_testing_dendrite_capacity(False)

# ----------------------------
# Build + initialize PAI
# ----------------------------
base = TabModel(n_num=X_num.shape[1], cat_sizes=cat_sizes).to(device)
model = UPA.initialize_pai(base).to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)

# PerforatedAI tracker method name differs across versions
if hasattr(GPA.pai_tracker, "set_optimizer_instance"):
    GPA.pai_tracker.set_optimizer_instance(optimizer)
else:
    GPA.pai_tracker.setOptimizerInstance(optimizer)

scaler_amp = torch.cuda.amp.GradScaler(enabled=torch.cuda.is_available())

def train_steps(steps=TRAIN_STEPS_PER_OUTER):
    model.train()
    it = iter(dl_tr)
    for _ in range(steps):
        try:
            xn, xc, yy = next(it)
        except StopIteration:
            it = iter(dl_tr)
            xn, xc, yy = next(it)
        xn, xc, yy = xn.to(device), xc.to(device), yy.to(device)
        optimizer.zero_grad(set_to_none=True)
        with torch.amp.autocast(device_type="cuda", enabled=torch.cuda.is_available()):
            logits = model(xn, xc)
            loss = F.binary_cross_entropy_with_logits(logits, yy)
        scaler_amp.scale(loss).backward()
        scaler_amp.step(optimizer)
        scaler_amp.update()

@torch.no_grad()
def val_auc():
    model.eval()
    ps, ys = [], []
    for xn, xc, yy in dl_va:
        xn, xc = xn.to(device), xc.to(device)
        logits = model(xn, xc)
        ps.append(torch.sigmoid(logits).cpu().numpy())
        ys.append(yy.numpy())
    p = np.concatenate(ps)
    t = np.concatenate(ys)
    return float(roc_auc_score(t, p)), p, t

# ----------------------------
# PAI outer loop (hard-capped)
# ----------------------------
best = 0.0
no_imp = 0

for outer in range(MAX_OUTER):
    train_steps()
    auc, _, _ = val_auc()

    if auc > best + IMPROVE_EPS:
        best = auc
        no_imp = 0
    else:
        no_imp += 1

    model, restructured, complete = GPA.pai_tracker.add_validation_score(auc, model)
    model = model.to(device)

    print(f"[FAST6] outer={outer:02d} auc={auc:.6f} best={best:.6f} no_imp={no_imp} restr={int(restructured)} complete={complete}")

    # If PAI completes, stop
    if complete:
        break

    # If restructured, reinit optimizer (as required)
    if restructured:
        optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
        if hasattr(GPA.pai_tracker, "set_optimizer_instance"):
            GPA.pai_tracker.set_optimizer_instance(optimizer)
        else:
            GPA.pai_tracker.setOptimizerInstance(optimizer)

    # Plateau stop
    if no_imp >= NO_IMPROVE_LIMIT:
        print("Stopping early due to plateau.")
        break

print("Final best val AUC:", best)

# ----------------------------
# Predict test + save submission
# ----------------------------
@torch.no_grad()
def predict_test():
    ds_te = TabDS(T_num, T_cat, y=None)
    dl_te = DataLoader(ds_te, batch_size=BATCH_EVAL, shuffle=False, num_workers=0)
    preds = []
    model.eval()
    for xn, xc in dl_te:
        xn, xc = xn.to(device), xc.to(device)
        preds.append(torch.sigmoid(model(xn, xc)).cpu().numpy())
    return np.concatenate(preds).astype(np.float32)

test_pred = predict_test()

# Safety flip if something is badly inverted locally (rare, but avoids 0.28 disasters)
auc_now, p_va, t_va = val_auc()
if auc_now < 0.5:
    print("AUC<0.5 detected -> flipping probabilities")
    test_pred = 1.0 - test_pred

submission = pd.DataFrame({"id": test_df["id"].values, "Heart Disease": test_pred})
submission.to_csv(OUT_PATH, index=False)
print("Saved:", OUT_PATH)
print(submission.head())
```

```python
# =========================================================
# ~6 MINUTES Colab Run (T4): PerforatedAI Dendrites Tabular NN
# - 1-fold (StratifiedShuffleSplit)
# - hard stop: MAX_OUTER=10, TRAIN_STEPS=140
# - skips PAI interactive ipdb prompts
# - GPU + AMP
# - writes /content/submission.csv
# =========================================================

!pip -q install git+https://github.com/PerforatedAI/PerforatedAI.git
!pip -q install scikit-learn pandas numpy

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import roc_auc_score

from perforatedai import globals_perforatedai as GPA
from perforatedai import utils_perforatedai as UPA

# ----------------------------
# Config (6-min budget)
# ----------------------------
TRAIN_PATH = "/content/train.csv"
TEST_PATH  = "/content/test.csv"
OUT_PATH   = "/content/submission.csv"

ID_COL = "id"
TARGET_COL = "Heart Disease"

SEED = 2042

VAL_FRAC = 0.12
MAX_OUTER = 10              # hard cap for time
TRAIN_STEPS_PER_OUTER = 140 # hard cap for time
BATCH_TRAIN = 16384
BATCH_EVAL  = 32768

LR = 1e-3
# PerforatedAI recommends no weight decay; keep it 0 for speed + fewer prompts
WEIGHT_DECAY = 0.0

# Plateau stop: if best doesn't improve for N outers, exit
NO_IMPROVE_LIMIT = 4
IMPROVE_EPS = 3e-5

torch.manual_seed(SEED)
np.random.seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device, torch.cuda.get_device_name(0) if torch.cuda.is_available() else "")

# ----------------------------
# Load data
# ----------------------------
train_df = pd.read_csv(TRAIN_PATH)
test_df  = pd.read_csv(TEST_PATH)

y = train_df[TARGET_COL].map({"Presence": 1, "Absence": 0}).astype(np.int64).values
X_df = train_df.drop(columns=[ID_COL, TARGET_COL]).copy()
T_df = test_df.drop(columns=[ID_COL]).copy()

# Categorical columns (heart dataset standard)
cat_cols = ["Sex","Chest pain type","FBS over 120","EKG results","Exercise angina","Slope of ST","Number of vessels fluro","Thallium"]
cat_cols = [c for c in cat_cols if c in X_df.columns]
num_cols = [c for c in X_df.columns if c not in cat_cols]

# Encode categoricals jointly to keep consistent codes
if len(cat_cols) > 0:
    all_cat = pd.concat([X_df[cat_cols], T_df[cat_cols]], axis=0, ignore_index=True)
    for c in cat_cols:
        all_cat[c] = all_cat[c].astype("int64")
        uniq = np.sort(all_cat[c].unique())
        mp = {v:i for i,v in enumerate(uniq)}
        all_cat[c] = all_cat[c].map(mp).astype("int64")

    X_cat = all_cat.iloc[:len(X_df)][cat_cols].values.astype(np.int64)
    T_cat = all_cat.iloc[len(X_df):][cat_cols].values.astype(np.int64)
    cat_sizes = [int(all_cat[c].nunique()) for c in cat_cols]
else:
    X_cat = np.zeros((len(X_df), 0), dtype=np.int64)
    T_cat = np.zeros((len(T_df), 0), dtype=np.int64)
    cat_sizes = []

# Scale numerics
scaler = StandardScaler()
X_num = scaler.fit_transform(X_df[num_cols].values.astype(np.float32)).astype(np.float32)
T_num = scaler.transform(T_df[num_cols].values.astype(np.float32)).astype(np.float32)

# ----------------------------
# Dataset
# ----------------------------
class TabDS(Dataset):
    def __init__(self, Xn, Xc, y=None):
        self.Xn = torch.from_numpy(Xn)
        self.Xc = torch.from_numpy(Xc)
        self.y  = None if y is None else torch.from_numpy(y.astype(np.float32))
    def __len__(self): return self.Xn.shape[0]
    def __getitem__(self, i):
        if self.y is None:
            return self.Xn[i], self.Xc[i]
        return self.Xn[i], self.Xc[i], self.y[i]

# ----------------------------
# Strong but fast model
# ----------------------------
class TabModel(nn.Module):
    def __init__(self, n_num, cat_sizes, emb_dim=8, dropout=0.15):
        super().__init__()
        self.embs = nn.ModuleList([nn.Embedding(s, min(emb_dim, max(2, s//2))) for s in cat_sizes])
        cat_out = sum(e.embedding_dim for e in self.embs)

        self.bn0 = nn.BatchNorm1d(n_num + cat_out)
        self.fc1 = nn.Linear(n_num + cat_out, 256)
        self.bn1 = nn.BatchNorm1d(256)
        self.fc2 = nn.Linear(256, 128)
        self.bn2 = nn.BatchNorm1d(128)
        self.fc3 = nn.Linear(128, 64)
        self.out = nn.Linear(64, 1)
        self.drop = nn.Dropout(dropout)

    def forward(self, x_num, x_cat):
        if len(self.embs) > 0:
            embs = [emb(x_cat[:, i]) for i, emb in enumerate(self.embs)]
            x = torch.cat([x_num] + embs, dim=1)
        else:
            x = x_num
        x = self.bn0(x)
        x = self.drop(F.relu(self.bn1(self.fc1(x))))
        x = self.drop(F.relu(self.bn2(self.fc2(x))))
        x = self.drop(F.relu(self.fc3(x)))
        return self.out(x).squeeze(-1)

# ----------------------------
# One fast stratified split
# ----------------------------
sss = StratifiedShuffleSplit(n_splits=1, test_size=VAL_FRAC, random_state=SEED)
tr_idx, va_idx = next(sss.split(X_num, y))

ds_tr = TabDS(X_num[tr_idx], X_cat[tr_idx], y[tr_idx])
ds_va = TabDS(X_num[va_idx], X_cat[va_idx], y[va_idx])

dl_tr = DataLoader(ds_tr, batch_size=BATCH_TRAIN, shuffle=True, num_workers=0)
dl_va = DataLoader(ds_va, batch_size=BATCH_EVAL, shuffle=False, num_workers=0)

# ----------------------------
# PerforatedAI config: avoid prompts
# ----------------------------
# Skip "unwrapped modules" confirmations (BN + Embeddings)
if hasattr(GPA, "pc"):
    if hasattr(GPA.pc, "set_unwrapped_modules_confirmed"):
        GPA.pc.set_unwrapped_modules_confirmed(True)
    # Track unwrapped modules so PAI won't complain
    try:
        GPA.pc.get_module_names_to_track().add("BatchNorm1d")
        GPA.pc.get_module_names_to_track().add("Embedding")
    except Exception:
        pass

    # If you keep weight_decay=0, no warning; but accept anyway to avoid breakpoints
    if hasattr(GPA.pc, "set_weight_decay_accepted"):
        GPA.pc.set_weight_decay_accepted(True)

    # Disable test/debug capacity behavior
    if hasattr(GPA.pc, "set_testing_dendrite_capacity"):
        GPA.pc.set_testing_dendrite_capacity(False)

# ----------------------------
# Build + initialize PAI
# ----------------------------
base = TabModel(n_num=X_num.shape[1], cat_sizes=cat_sizes).to(device)
model = UPA.initialize_pai(base).to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)

# PerforatedAI tracker method name differs across versions
if hasattr(GPA.pai_tracker, "set_optimizer_instance"):
    GPA.pai_tracker.set_optimizer_instance(optimizer)
else:
    GPA.pai_tracker.setOptimizerInstance(optimizer)

scaler_amp = torch.cuda.amp.GradScaler(enabled=torch.cuda.is_available())

def train_steps(steps=TRAIN_STEPS_PER_OUTER):
    model.train()
    it = iter(dl_tr)
    for _ in range(steps):
        try:
            xn, xc, yy = next(it)
        except StopIteration:
            it = iter(dl_tr)
            xn, xc, yy = next(it)
        xn, xc, yy = xn.to(device), xc.to(device), yy.to(device)
        optimizer.zero_grad(set_to_none=True)
        with torch.amp.autocast(device_type="cuda", enabled=torch.cuda.is_available()):
            logits = model(xn, xc)
            loss = F.binary_cross_entropy_with_logits(logits, yy)
        scaler_amp.scale(loss).backward()
        scaler_amp.step(optimizer)
        scaler_amp.update()

@torch.no_grad()
def val_auc():
    model.eval()
    ps, ys = [], []
    for xn, xc, yy in dl_va:
        xn, xc = xn.to(device), xc.to(device)
        logits = model(xn, xc)
        ps.append(torch.sigmoid(logits).cpu().numpy())
        ys.append(yy.numpy())
    p = np.concatenate(ps)
    t = np.concatenate(ys)
    return float(roc_auc_score(t, p)), p, t

# ----------------------------
# PAI outer loop (hard-capped)
# ----------------------------
best = 0.0
no_imp = 0

for outer in range(MAX_OUTER):
    train_steps()
    auc, _, _ = val_auc()

    if auc > best + IMPROVE_EPS:
        best = auc
        no_imp = 0
    else:
        no_imp += 1

    model, restructured, complete = GPA.pai_tracker.add_validation_score(auc, model)
    model = model.to(device)

    print(f"[FAST6] outer={outer:02d} auc={auc:.6f} best={best:.6f} no_imp={no_imp} restr={int(restructured)} complete={complete}")

    # If PAI completes, stop
    if complete:
        break

    # If restructured, reinit optimizer (as required)
    if restructured:
        optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
        if hasattr(GPA.pai_tracker, "set_optimizer_instance"):
            GPA.pai_tracker.set_optimizer_instance(optimizer)
        else:
            GPA.pai_tracker.setOptimizerInstance(optimizer)

    # Plateau stop
    if no_imp >= NO_IMPROVE_LIMIT:
        print("Stopping early due to plateau.")
        break

print("Final best val AUC:", best)

# ----------------------------
# Predict test + save submission
# ----------------------------
@torch.no_grad()
def predict_test():
    ds_te = TabDS(T_num, T_cat, y=None)
    dl_te = DataLoader(ds_te, batch_size=BATCH_EVAL, shuffle=False, num_workers=0)
    preds = []
    model.eval()
    for xn, xc in dl_te:
        xn, xc = xn.to(device), xc.to(device)
        preds.append(torch.sigmoid(model(xn, xc)).cpu().numpy())
    return np.concatenate(preds).astype(np.float32)

test_pred = predict_test()

# Safety flip if something is badly inverted locally (rare, but avoids 0.28 disasters)
auc_now, p_va, t_va = val_auc()
if auc_now < 0.5:
    print("AUC<0.5 detected -> flipping probabilities")
    test_pred = 1.0 - test_pred

submission = pd.DataFrame({"id": test_df["id"].values, "Heart Disease": test_pred})
submission.to_csv(OUT_PATH, index=False)
print("Saved:", OUT_PATH)
print(submission.head())
```
